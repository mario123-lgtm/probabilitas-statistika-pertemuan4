```python
import pandas as pd
```

```python
datafar = pd.read_excel("farid.xlsx")
print(datafar)
```

```python
datafreq = pd.crosstab(index=datafar["Grade"], columns="Frekuensi")
print(datafreq)
```

```python
datafreq.plot()
```

```python
ax = datafreq.plot(kind="bar", title="Frekuensi Grade Mahasiswa")
ax.set_xlabel("Grade")
ax.set_ylabel("Frekuensi")
```

```python
datafreq.plot(
    kind="pie",
    y="Frekuensi",
    autopct="%1.0f%%",
    legend=False,
    title="Persentase Grade Mahasiswa"
)
```

```python
datafar["Nilai"] = pd.to_numeric(datafar["Nilai"], errors="coerce")

dt = datafar["Nilai"]

stats = dt.describe()

stats["Standard Error"] = dt.sem()
stats["Median"] = dt.median()
stats["Mode"] = dt.mode().iloc[0]
stats["Variance"] = dt.var()
stats["Range"] = dt.max() - dt.min()
stats["Skewness"] = dt.skew()
stats["Kurtosis"] = dt.kurtosis()

print(stats)
```

```python
Q1 = dt.quantile(0.25)
Q3 = dt.quantile(0.75)
IQR = Q3 - Q1

batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR

outlier = datafar[
    (datafar["Nilai"] < batas_bawah) |
    (datafar["Nilai"] > batas_atas)
]

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Batas Bawah =", batas_bawah)
print("Batas Atas =", batas_atas)
print("Jumlah Outlier =", len(outlier))
print(outlier)
```
